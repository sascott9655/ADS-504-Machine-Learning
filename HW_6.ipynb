##1.1. Video capturing

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
```

1. These are the imports needed to capture video. Display is able to show data that is more complex than text and numbers such as sound and video.
2. Javascript is a programming language that enables us data scientists to use API fetch calls to connect our the software to hardware (THIS COULD BE WRONG BUT INITIAL INTUITION). It also is a programming known for creating interactive elements on a webpage.
The google.colab.output import eval_js evaluates the javascript and is able to compile Javascript code in a Python environment.
3. The b64decode import allows us to decode string data that can be understood by the computer and to convert it to an image.


```python
def take_photo(filename='photo.jpg', quality=0.8): #Create a method that takes two parameters the file of the photo and the
#quality which is set to 0.8 (reducing quality of image makes the image file less small).

  js = Javascript(''' // now we are coding in Javascript
    async function takePhoto(quality) { // async functions are used so that other Javascript can run while a longer
    //time taking task can run in the background. We are creating a function within a function.
    //The takePhoto function takes the quality parameter of the photo.

      const div = document.createElement('div'); // create a div element from HTML
      const capture = document.createElement('button'); // create a clickable interactive button like in a website program
      capture.textContent = 'Capture'; // have the button say Capture inside the button box
      div.appendChild(capture); // append that button functionality and button inside the div

      const video = document.createElement('video'); // create a video element object
      video.style.display = 'block'; // display the video object so that it is visible to the webpage
      const stream = await navigator.mediaDevices.getUserMedia({video: true}); // the await keyword prioritizes itself to run while
      //pausing the async function. This line of code gets the user's camera on the computer connecting software and hardware.

      document.body.appendChild(div); // append that div the the body function of the webpage
      div.appendChild(video); // append the video element inside the div
      video.srcObject = stream; // connect the stream variable which retrieves your camera and put it in the video object
      await video.play(); // play your camera
      google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true); // Resize the output to fit the video element.
      await new Promise((resolve) => capture.onclick = resolve); // Wait for Capture to be clicked.
      const canvas = document.createElement('canvas'); // create a canvas element which renders
      //2D shapes, 3D graphics, animations, and game visuals directly onto a webpage using JavaScript - Google Gemini
      canvas.width = video.videoWidth; // set the width of the canvas to the video width
      canvas.height = video.videoHeight; // set the height of the canvas to the video height
      canvas.getContext('2d').drawImage(video, 0, 0); // draw a 2d image of the video object on the webpaga
      stream.getVideoTracks()[0].stop(); // stops the video camera turns off camera
      div.remove(); // remove the div, which removes the use of the camera
      return canvas.toDataURL('image/jpeg', quality); // return the image and save it to your computer
      }
      ''')
  #Back to python code
  display(js) # show the javascript function working
  data = eval_js('takePhoto({})'.format(quality)) #store the taken photo in data using the async JS function
  binary = b64decode(data.split(',')[1]) #remove heading of image which is why we split on comma. The 1 index is the image heading.
  with open(filename, 'wb') as f:  #save the binary heading as the filename
      f.write(binary) #write is to output the binary into the f
  return filename #return the filename
```
